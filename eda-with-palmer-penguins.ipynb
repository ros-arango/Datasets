##  Univatient Analysis 

```r
ggplot(penguins )+
    geom_histogram(aes(x = body_mass_g),
                  fill = "red",
                  color = "black")+
    labs(title = "Highest Body Mass of Penguins",
         caption = "source : Palmer Penguins"
    )
    
```

```r
ggplot(penguins)+
    geom_density(aes(x = flipper_length_mm ),
                  color = "red")+
    labs(title = "Highest Flipper Lenght of Penguins",
         caption = "source : Palmer Penguins"
        )
```

## Multivarient Analysis

```r
ggplot(penguins)+
    geom_point(aes(x = bill_length_mm, y = flipper_length_mm),
                  color = "red") +
    geom_smooth(aes(x = bill_length_mm, y = flipper_length_mm))+
    labs(title = " bill_length_mm VS bill_depth_mm ",
         caption = "source : Palmer Penguins"
        ) 
```

```r
ggplot(penguins)+
    geom_col(aes(x = bill_length_mm, y = flipper_length_mm),
                  color = "red",
                  fill = "black"
                ) +
    geom_smooth(aes(x = bill_length_mm, y = flipper_length_mm))+
    labs(title = " bill_length_mm VS bill_depth_mm ",
         caption = "source : Palmer Penguins"
        ) 
```

## Questions and Answers 


### 1. Relationship of bill_length_mm and flipper_length_mm with Scatter plot of species -  penguins.

```r
ggplot(penguins)+
    geom_point(aes(x = bill_length_mm, y = flipper_length_mm, color = species))+
    labs(title = " bill_length_mm VS bill_depth_mm ",
         subtitle = "Differentiating with penguis",
         caption = "source : Palmer Penguins"
        ) 
```

```r
ggplot(penguins)+
    geom_point(aes(x = bill_length_mm, y = flipper_length_mm, color = species))+
    labs(title = " bill_length_mm VS bill_depth_mm ",
         subtitle = "Differentiating with penguis gender",
         caption = "source : Palmer Penguins"
        ) +
    facet_wrap(~sex)
```

```r
ggplot(penguins)+
    geom_point(aes(x = bill_length_mm, y = flipper_length_mm, color = species))+
    labs(title = " bill_length_mm VS bill_depth_mm ",
         subtitle = "Differentiating with penguis species",
         caption = "source : Palmer Penguins"
        ) +
    facet_wrap(~species)
```

### 2. Bodymass weight with respect to gender of the penguins 

```r
ggplot(penguins)+
    geom_col(aes(x = bill_length_mm, y = flipper_length_mm, color = species))+
    labs(title = " bill_length_mm vs bill_depth_mm ",
         subtitle = "Differentiating with penguis specise",
         caption = "source : Palmer Penguins"
        ) 
```

### 3.Density of Penguins in scatter plot 

```r
ggplot(penguins)+
    geom_count(aes(x = bill_length_mm, y = flipper_length_mm, color = species))+
    labs(title = " bill_length_mm VS bill_depth_mm ",
         subtitle = "Differentiating with penguis",
         caption = "source : Palmer Penguins"
        ) 
```

### 4. Number of penguins over years

```r
penguins %>%
filter(sex != 'NA') %>%
ggplot()+
    geom_bar(aes(x = year, fill = species))+
    facet_wrap(~sex)
```
